```python
from gitma import CatmaProject, _metrics

catma_project = CatmaProject(
    projects_directory='../projects/',
    project_name='GitMA_Demo_Project',
)
```

```python
ac_1 = catma_project.ac_dict["ac_1"]
ac_2 = catma_project.ac_dict["ac_2"]
catma_project.calculate_cohens_kappa(ac1_name_or_inst=ac_1, ac2_name_or_inst=ac_2)
catma_project.calculate_scotts_pi(ac1_name_or_inst=ac_1, ac2_name_or_inst=ac_2)
catma_project.calculate_krippendorffs_alpha_for_two_collections(ac1_name_or_inst=ac_1, ac2_name_or_inst=ac_2)
print(catma_project.annotation_collections)
catma_project.calculate_krippendorffs_alpha_for_multiple(catma_project.annotation_collections, level='tag')

catma_project.get_iaa(ac_1, ac_2)

# Default pairwise IAA data
_metrics.get_iaa_data(ac_1, ac_2)
# Multiple IAA data (all collections)
list(_metrics.get_iaa_data_multiple(catma_project.annotation_collections))
# Combination IAA data (all collection pairs)
# This does not work on demo project since gold ac has no annotations
_metrics.get_iaa_data_combination([ac_1, ac_2])
```
